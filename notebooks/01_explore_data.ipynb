```python
import pandas as pd

```

```python
from pathlib import Path

relative_path = Path("data/processed/amazon_cellphones_clean.csv")
project_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / relative_path).is_file()),
    None,
)

if project_root is None:
    raise FileNotFoundError(
        f"Could not find {relative_path} from {Path.cwd()}. "
        "Check that the file exists and that the data preprocessing step has run."
    )

df = pd.read_csv(project_root / relative_path)
```

```python
df.head()
```

```python
print("Shape:", df.shape)

print("\n=== COLUMNS ===")
print(df.columns.tolist())

print("\n=== RATING ===")
print(df["overall"].value_counts().sort_index())

print("\n=== ASIN ===")
print("Unique products:", df["asin"].nunique())
print(df["asin"].value_counts().head(10))

print("\n=== YEAR ===")
print(df["year"].value_counts().sort_index())

print("\n=== SAMPLE ===")
display(df[["asin", "reviewText", "summary", "overall", "reviewTime", "review_date", "year"]].head(10))
```

```python
checkpoint_path = (
    project_root
    / "data"
    / "processed"
    / "explore_amazon_cellphones.pkl"
)

df.to_pickle(checkpoint_path)

print("Checkpoint saved:", checkpoint_path)
print("Shape:", df.shape)
```
